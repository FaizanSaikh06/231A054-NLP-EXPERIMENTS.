```python

text = '''Hey everyone!!! 👋

I'm REALLY excited 😊 because I've just completed my first Machine Learning project on 23rd July, 2026. The project analyzed over 12,500 customer reviews collected from https://www.example.com and another dataset downloaded from www.kaggle.com.

Unfortunately, some reviews weren't useful—they contained only words like "the", "is", "and", "of", "to", etc. Others had strange symbols such as @#$%^&*, repeated punctuation!!!!!!, HTML tags like <br>, and emojis 😂😂😂.

You can contact me at vijay.shanker@eng.rizvi.edu.in or support123@company.co.in if you'd like the dataset. Follow us on Twitter @AI_ResearchLab and use hashtags like #MachineLearning #NLP #DataScience.

The model's accuracy increased from 82.5% to 91.8%, while the training time decreased by nearly 35%. It processed 1,25,000 records in less than 2.5 hours and consumed around 3.7 GB of RAM.

Students were studying, analyzing, processing, predicting, running, writing, and comparing different algorithms. The researchers discovered better techniques after performing extensive experiments and evaluating multiple models.

Don't worry if the results aren't perfect—you're learning! Let's improve them together. I'll upload the final report by 30/07/2026.

Thanks!!!'''
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

print(text.lower())

text_lower = text.lower()

print("Faizan Sheikh", "UIN: 231A054\n")
```

```python
import re

# First, let's remove URLs from text_lower to create text_url
url_pattern = r'https?:\/\/\S+|www\.\S+'
text_url = re.sub(url_pattern, " ", text_lower)

# Now, define the pattern to remove email addresses
pattern = r'\b\S+@\S+\b'

text_emails = re.sub(pattern, " ", text_url)

print(text_emails)
```

```python

pattern = r'\d[\d,.]*%?'

text_percentage = re.sub(pattern, " ", text_emails)

print(text_percentage)
```

```python

import emoji

text_emoji = emoji.replace_emoji(text_percentage, replace="")

print(text_emoji)
```

```python

import nltk

nltk.download('punkt_tab')
```

```python

from nltk import word_tokenize

tokenized = word_tokenize(text_emoji)

print("Tokenized words:", tokenized)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

import string

punctuation = list(string.punctuation)

print(punctuation)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

print("With punctuation:", tokenized)
print()

text_punct = []

for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)

print("Without punctuation:", text_punct)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

print(stopwords.words("english"))
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

stop_words = set(stopwords.words("english"))

text_clean2 = []

for word in text_punct:
    if word.lower() not in stop_words:
        text_clean2.append(word.lower())

print(text_clean2)
```

```python

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in text_clean2]

print("Stemmed words:", stemmed_words)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

import nltk

nltk.download('wordnet')
```

```python

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in text_clean2
]

print("Lemmatized words:", lemmatized_words)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

tokens_before_stop_words = len(text_punct)
tokens_after_stop_words = len(text_clean2)
tokens_after_lemmatization = len(lemmatized_words)

print(
    f"Number of tokens before stop word removal "
    f"(after punctuation): {tokens_before_stop_words}"
)

print(
    f"Number of tokens after stop word removal: "
    f"{tokens_after_stop_words}"
)

print(
    f"Number of tokens after lemmatization: "
    f"{tokens_after_lemmatization}"
)

print("-" * 30)
```

```python

vocab_before_stop_words = len(set(text_punct))
vocab_after_stop_words = len(set(text_clean2))
vocab_after_lemmatization = len(set(lemmatized_words))

print(
    f"Vocabulary size before stop word removal "
    f"(after punctuation): {vocab_before_stop_words}"
)

print(
    f"Vocabulary size after stop word removal: "
    f"{vocab_after_stop_words}"
)

print(
    f"Vocabulary size after lemmatization: "
    f"{vocab_after_lemmatization}"
)

print("-" * 30)
```

```python

print("Faizan Sheikh", "UIN: 231A054\n")

# Regenerate stemmed words for comparison
stemmed_words = [stemmer.stem(word) for word in text_clean2]

print("Stemmed words:", stemmed_words)

print("\nLemmatized words:", lemmatized_words)
```
