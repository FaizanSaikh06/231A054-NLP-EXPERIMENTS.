```python

import nltk

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
```

```python

from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python

print("Faizan Sheikh - 231A054\n")

text_corpus = "The walker is walking slowly."

print(f"Input Text: '{text_corpus}'")
```

```python

preprocessed_text = text_corpus.lower()

print(f"Lowercase Text: '{preprocessed_text}'")
```

```python

translator = str.maketrans('', '', string.punctuation)

preprocessed_text = preprocessed_text.translate(translator)

print(f"Preprocessed Text: '{preprocessed_text}'")
```

```python
import nltk
nltk.download('punkt_tab')

tokens = word_tokenize(preprocessed_text)

print(f"Tokens: {tokens}")
```

```python

lemmatizer = WordNetLemmatizer()

morpheme_analysis = {}
```

```python

word_walker = "walker"

# Manually assign the conceptual root
root_walker = "walk"

suffix_walker = "-er" if word_walker.endswith("er") else ""

morpheme_analysis[word_walker] = {
    "root": root_walker,
    "suffix": suffix_walker,
    "type": "derivational"
}
```

```python

word_walking = "walking"

root_walking = lemmatizer.lemmatize(word_walking, pos='v')

suffix_walking = "-ing" if word_walking.endswith("ing") else ""

morpheme_analysis[word_walking] = {
    "root": root_walking,
    "suffix": suffix_walking,
    "type": "inflectional"
}
```

```python

print("Morpheme Analysis:")

for word, analysis in morpheme_analysis.items():
    print(
        f"  '{word}': Root = '{analysis['root']}', "
        f"Suffix = '{analysis['suffix']}' "
        f"({analysis['type']} morphology)"
    )
```

```python

generated_forms = {"walk": []}

# Forms for 'walk'
base_form = "walk"

generated_forms[base_form].append(base_form)
generated_forms[base_form].append("walking")
generated_forms[base_form].append("walked")
generated_forms[base_form].append("walker")

print("Generated Word Forms:")

for root, forms in generated_forms.items():
    print(f"  '{root}' (base form) -> {', '.join(forms)}")
```

```python

print("Faizan Sheikh - 231A054")

print("\n--- Summary of Findings ---")

print("Observation:")
print("  'walk' can have different word forms based on morphological rules.")
print("  'walker' is derived through derivational morphology (root 'walk' + suffix '-er').")
print("  'walking' is a result of inflectional morphology (root 'walk' + suffix '-ing').")
```

```python

print("\nOutput:")

print(f"Words Analyzed: {', '.join(morpheme_analysis.keys())}")
```

```python

print("Morphemes Identified:")

for word, analysis in morpheme_analysis.items():
    print(
        f"  '{word}': Root = '{analysis['root']}', "
        f"Suffix = '{analysis['suffix']}'"
    )
```

```python

all_generated_forms = set()

for forms in generated_forms.values():
    all_generated_forms.update(forms)

print(
    f"Generated Word Forms: "
    f"{', '.join(sorted(all_generated_forms))}"
)
```
